```python
from clickhouse_driver import Client
import pandas as pd
```

```python
client = Client(
    host='',
    user='default',
    password='',
    secure=True
)
```

```python
print(client.execute('SELECT 1'))
```

```python
query = """
SELECT *
FROM helloworld.`trips`
LIMIT 10
"""

data = client.execute(query)
```

```python
pd.DataFrame(data)
```

```python
columns = [col[0] for col in client.execute(
    "DESCRIBE TABLE helloworld.`trips`"
)]
```

```python

df = pd.DataFrame(data, columns=columns)
df.head()
```

```python
metrics= {
    "total_trips": len(df),
    "total_revenue": df['fare_amount'].sum(),
    "avg_fare": df['fare_amount'].mean(),
    "avg_distance": df['trip_distance'].mean(),
    "avg_passengers": df['passenger_count'].mean()
}
```

```python
metrics
```

```python
list(metrics.items())
```

```python
metrics_df=pd.DataFrame(
    list(metrics.items()),
    columns=["metric_name","metric_value"]
)
```

```python
metrics_df
```

```python
client.execute("""
CREATE TABLE IF NOT EXISTS helloworld.taxi_metrics
(
    metric_name String,
    metric_value Float64,
    created_at DateTime DEFAULT now()
)
ENGINE = MergeTree
ORDER BY metric_name
""")
```

```python
client.execute(
    "INSERT INTO helloworld.taxi_metrics (metric_name, metric_value) VALUES",
    metrics_df.to_dict("records")
)
```
