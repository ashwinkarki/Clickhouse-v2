```python
CREATE TABLE events (
    timestamp DateTime,
    user_id UInt64,
    event_type LowCardinality(String)
)
ENGINE = MergeTree()
PARTITION BY toYYYYMM(timestamp)
ORDER BY (timestamp, user_id);
```

```python
INSERT INTO events
FORMAT CSV
```
